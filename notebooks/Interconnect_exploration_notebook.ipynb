# Telecom

A empresa de telecomunicações **Interconnect** busca **prever a taxa de cancelamento de seus clientes (churn)**.

O objetivo é antecipar quais usuários podem deixar o serviço para oferecer **códigos promocionais** e **planos especiais** que incentivem sua permanência.

A equipe de marketing da Interconnect reuniu informações detalhadas sobre seus clientes, incluindo **dados pessoais**, **características dos planos contratados** e **tipos de serviços utilizados**.

---

## 💼 Serviços da Interconnect

A Interconnect oferece principalmente **dois tipos de serviços**:

1. **Telefonia fixa:** permite conectar várias linhas simultaneamente.
2. **Internet:** disponível por meio de uma linha telefônica (*DSL — linha de assinante digital*) ou através de **fibra óptica**.

Além disso, a empresa oferece **serviços complementares** que aprimoram a experiência do cliente:

- 🔒 **Segurança na Internet:** software antivírus (*ProteçãoDeDispositivo*) e bloqueador de sites maliciosos (*SegurançaOnline*).
- ☎️ **Suporte técnico:** assistência personalizada para resolver problemas técnicos (*SuporteTécnico*).
- ☁️ **Armazenamento em nuvem e backup de dados:** serviço de *BackupOnline*.
- 📺 **Entretenimento digital:** acesso a *StreamingTV* e *StreamingFilmes* (catálogo de séries e filmes).

Os clientes podem optar por **pagamentos mensais** ou **contratos de 1 ou 2 anos**, com múltiplas formas de pagamento disponíveis.

Ao final de cada transação, eles recebem uma **fatura eletrônica**.

---

## 📊 Descrição dos Dados

Os dados vêm de diferentes fontes e estão organizados nos seguintes arquivos:

- `contract.csv` — informações contratuais do cliente.
- `personal.csv` — dados pessoais do cliente.
- `internet.csv` — detalhes sobre os serviços de Internet contratados.
- `phone.csv` — detalhes sobre os serviços telefônicos contratados.

Em cada arquivo, a coluna **`customerID`** identifica de forma única cada cliente.

As informações dos contratos são **válidas a partir de 1º de fevereiro de 2020**.
