# Indústria – Steelproof

A planta siderúrgica **Steelproof** busca **otimizar seus custos de produção** reduzindo o consumo de energia durante a etapa de **processamento do aço**.

Seu objetivo será **desenvolver um modelo capaz de prever a temperatura do metal**, ajudando a planta a tomar decisões mais eficientes e sustentáveis em seu processo industrial.

---

## 🔥 Etapas do Processamento do Aço

O aço é processado em uma **panela metalúrgica** com capacidade aproximada de **100 toneladas**.

Para suportar as altas temperaturas, o interior da panela é revestido com **tijolos refratários**.

1. **Fusão e aquecimento:**
    
    O aço fundido é despejado na panela e aquecido até atingir a temperatura desejada por meio de **eletrodos de grafite** localizados na tampa.
    
2. **Dessulfurização:**
    
    O enxofre é removido do metal através de um processo químico especializado, aumentando sua pureza e qualidade.
    
3. **Ajuste químico:**
    
    A composição do aço é corrigida pela adição de **ligas metálicas**, em forma de lingotes adicionados a partir de um búnquer ou por meio de um **alimentador de arame**. Durante esse processo, também são retiradas **amostras** para análise.
    
4. **Controle de temperatura e composição:**
    
    Antes da adição dos elementos de liga, são feitas **medições de temperatura** e análises de **composição química**. Em seguida, o aço é aquecido por alguns minutos adicionais, os elementos necessários são adicionados e a liga é **purgada com gás inerte**.
    
5. **Homogeneização:**
    
    Após a purga, a liga é **agitada e a temperatura é medida novamente**. Esse ciclo se repete várias vezes até que se alcance a **composição química desejada** e a **temperatura ideal de fusão**.
    

Por fim, o aço fundido é **refinado** ou transferido para uma **máquina de lingotamento contínuo**, onde se solidifica formando **placas** que serão utilizadas em etapas posteriores da manufatura.
